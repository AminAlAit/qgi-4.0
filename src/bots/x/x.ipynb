### Create Example `top_patterns.parquet`

```python
import pandas as pd
import os

print("Creating a mock 'top_patterns.parquet' for bot testing...")

source_file = r"/workspace/qgi\qgi-4.0\data\pattern_maker_output\Lebanon\patterns.parquet"
output_dir = r"/workspace/qgi\qgi-4.0\data\preprocessed_bot_data"
output_file = os.path.join(output_dir, "top_patterns.parquet")

try:
    df = pd.read_parquet(source_file)

    # CRUCIAL STEP: Add the missing SignificanceScore column
    # We'll create a dummy score based on correlation for testing.
    if "average_correlation" in df.columns:
        df["SignificanceScore"] = df["average_correlation"] * 0.8  # Example calculation
        df.sort_values("SignificanceScore", ascending=False, inplace=True)
    else:
        raise KeyError("'average_correlation' not found in source file.")

    # Ensure output directory exists
    os.makedirs(output_dir, exist_ok=True)

    # Save the new, correctly formatted file
    df.to_parquet(output_file, index=False)

    print(f"Successfully created mock data at: {output_file}")
    print(f"File contains {len(df)} rows.")
    print("\nTest data schema:")
    print(df.info())

except Exception as e:
    print(f"An error occurred: {e}")
```
