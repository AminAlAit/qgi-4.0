```python
try:
    # Fetch all available indicators using wbgapi
    indicators = wb.series.list()

    # Create lists to store the data
    indicator_ids = []
    indicator_names = []
    indicator_descriptions = []

    for indicator in indicators:  # indicators is already a list of dictionaries
        indicator_ids.append(indicator['id'])
        indicator_names.append(indicator['value'])

        # Extract the description from the 'sourceNote' field
        description = ""  # Initialize in case source or description are not found
        if 'sourceNote' in indicator:
            description = indicator['sourceNote']
        indicator_descriptions.append(description)

    # Create the DataFrame
    df_indicators = pd.DataFrame({
        'indicator_id': indicator_ids,
        'indicator_name': indicator_names,
        'indicator_description': indicator_descriptions
    })

    df_indicators.to_csv("world_bank_indicators.csv", index = False, encoding = "utf-8")
    print("DataFrame saved to world_bank_indicators.csv")

except Exception as e:
    print(f"An error occurred: {e}")
```
